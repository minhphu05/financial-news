```python
import glob

# Tìm tất cả file .jsonl trong thư mục hiện tại
input_files = sorted(glob.glob("/Users/kittnguyen/Documents/DS201_Finance/data/labeled/ner/raw/*.jsonl"))

output_file = "/Users/kittnguyen/Documents/DS201_Finance/data/labeled/ner/raw/merged_output.jsonl"

with open(output_file, "w", encoding="utf-8") as outfile:
    for fname in input_files:
        print("Đang gộp:", fname)
        with open(fname, "r", encoding="utf-8") as infile:
            for line in infile:
                line = line.strip()
                if line:  # bỏ dòng trống
                    outfile.write(line + "\n")

print("🎉 Đã gộp xong vào file:", output_file)
```

```python
import json

input_file = "/Users/kittnguyen/Documents/DS201_Finance/data/labeled/ner/raw/merged_output.jsonl"
output_file = "/Users/kittnguyen/Documents/DS201_Finance/data/labeled/ner/raw/output_ner.jsonl"

rows = []

# Đọc toàn bộ JSONL
with open(input_file, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        obj = json.loads(line)
        rows.append(obj)

# Sort theo id
rows.sort(key=lambda x: x["id"])

# Ghi ra file mới
with open(output_file, "w", encoding="utf-8") as f:
    for obj in rows:
        f.write(json.dumps(obj, ensure_ascii=False) + "\n")

print("🎉 Đã sắp xếp xong! File:", output_file)
```

```python
import json
import random

input_file = "merged_sorted.jsonl"

train_file = "train.jsonl"
dev_file = "dev.jsonl"
test_file = "test.jsonl"

# Đọc dữ liệu
data = []
with open(input_file, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if line:
            data.append(json.loads(line))

# Shuffle (xáo trộn)
random.shuffle(data)

# Tính số lượng cho mỗi tập
n = len(data)
n_train = int(n * 0.7)
n_dev = int(n * 0.15)
n_test = n - n_train - n_dev   # phần còn lại

train_data = data[:n_train]
dev_data = data[n_train:n_train + n_dev]
test_data = data[n_train + n_dev:]

# Hàm ghi file jsonl
def write_jsonl(filename, rows):
    with open(filename, "w", encoding="utf-8") as f:
        for obj in rows:
            f.write(json.dumps(obj, ensure_ascii=False) + "\n")

write_jsonl(train_file, train_data)
write_jsonl(dev_file, dev_data)
write_jsonl(test_file, test_data)

print("Done!")
print("Train:", len(train_data))
print("Dev:", len(dev_data))
print("Test:", len(test_data))
```

```python
import json

files = ["train.jsonl", "dev.jsonl", "test.jsonl"]

for filename in files:
    print("\n=== Kiểm tra file:", filename, "===")
    error_lines = []

    with open(filename, "r", encoding="utf-8") as f:
        for idx, line in enumerate(f, start=1):
            line = line.strip()
            if not line:
                continue

            try:
                obj = json.loads(line)
            except:
                error_lines.append((idx, "Không parse được JSON", line))
                continue

            keys = list(obj.keys())

            # dòng lỗi nếu:
            # - không có tokens
            # - không có tags
            # - hoặc key rỗng
            if ("tokens" not in obj) or ("tags" not in obj) or ("" in obj) or (len(keys) == 0):
                error_lines.append((idx, keys, line))

    if not error_lines:
        print("✔ Không có dòng lỗi!")
    else:
        print(f"❌ Tổng số dòng lỗi: {len(error_lines)}")
        for err in error_lines[:20]:  # in tối đa 20 dòng đầu cho dễ xem
            print("\n--- Lỗi ---")
            print("Dòng:", err[0])
            print("Keys:", err[1])
            print("Nội dung:", err[2])

        if len(error_lines) > 20:
            print(f"... còn {len(error_lines) - 20} lỗi nữa ...")
```

```python
import json

filename = "train.jsonl"

cleaned = []

with open(filename, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue

        obj = json.loads(line)

        # Nếu dòng có key rỗng "" → bỏ luôn, không giữ
        if "" in obj:
            continue

        cleaned.append(obj)

# Ghi đè file train.jsonl sạch
with open(filename, "w", encoding="utf-8") as f:
    for obj in cleaned:
        f.write(json.dumps(obj, ensure_ascii=False) + "\n")

print("🎉 Đã xóa các dòng lỗi chứa key rỗng trong train.jsonl!")
```

```python
import json

files = ["train.jsonl", "dev.jsonl", "test.jsonl"]

for filename in files:
    print("Đang xử lý:", filename)
    cleaned = []

    with open(filename, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue

            obj = json.loads(line)

            # XÓA KEY "sentence" nếu tồn tại
            if "sentence" in obj:
                obj.pop("sentence")

            cleaned.append(obj)

    # Ghi đè lại file
    with open(filename, "w", encoding="utf-8") as f:
        for obj in cleaned:
            f.write(json.dumps(obj, ensure_ascii=False) + "\n")

print("🎉 Đã xoá 'sentence' trong 3 file train, dev, test!")
```

```python
import json

files = ["train.jsonl", "dev.jsonl", "test.jsonl"]

for filename in files:
    print(f"\n===== Kiểm tra file: {filename} =====")
    errors = []

    with open(filename, "r", encoding="utf-8") as f:
        for idx, line in enumerate(f, start=1):
            line = line.strip()
            if not line:
                continue

            obj = json.loads(line)

            # Chỉ check khi có đủ keys
            if "tokens" not in obj or "tags" not in obj:
                continue

            t_len = len(obj["tokens"])
            g_len = len(obj["tags"])

            if t_len != g_len:
                errors.append((idx, t_len, g_len, line))

    if not errors:
        print("✔ Không có dòng mismatch tokens/tags")
    else:
        print(f"❌ Tổng số dòng mismatch: {len(errors)}")
        for e in errors[:20]:   # in tối đa 20 dòng cho dễ xem
            print("\n--- Lỗi ---")
            print("Dòng:", e[0])
            print("Số tokens:", e[1])
            print("Số tags:", e[2])
            print("Nội dung:", e[3])

        if len(errors) > 20:
            print(f"... còn {len(errors) - 20} lỗi nữa ...")
```

```python
import json

files = ["train.jsonl", "dev.jsonl", "test.jsonl"]

for filename in files:
    print(f"\n===== Xử lý file: {filename} =====")

    good_lines = []
    removed = 0

    with open(filename, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue

            try:
                obj = json.loads(line)
            except:
                removed += 1
                continue

            tokens = obj.get("tokens")
            tags = obj.get("tags")

            # Bỏ dòng không có tokens hoặc tags
            if tokens is None or tags is None:
                removed += 1
                continue

            if len(tokens) != len(tags):
                removed += 1
                continue

            good_lines.append(line)

    # Ghi lại file đã làm sạch
    with open(filename, "w", encoding="utf-8") as out:
        for gl in good_lines:
            out.write(gl + "\n")

    print(f"✔ Giữ lại {len(good_lines)} dòng")
    print(f"❌ Đã xoá {removed} dòng mismatch")
```
