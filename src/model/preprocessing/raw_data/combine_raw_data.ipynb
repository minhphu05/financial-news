# 1. Thêm index vào mỗi phần tử json trong cafeF_news.json

```python
import json
import os

# Dữ liệu sau khi được truy vấn từ MongoDB về chưa có index là số thứ tự, chỉ có _id chứa mã của riêng MongoDB
# --> Tạo thêm thuộc tính index
INPUT_FILE = "/Users/kittnguyen/Documents/DS201_Finance/data/raw/cafef_news_raw_final.json"
TEMP_FILE = INPUT_FILE + ".tmp"

# Load dữ liệu
with open(INPUT_FILE, "r", encoding="utf-8") as f:
    data = json.load(f)

# Thêm index
for idx, item in enumerate(data):
    item["index"] = idx

# Ghi ra file tạm
with open(TEMP_FILE, "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=4)

# Ghi đè file cũ bằng file tạm
os.replace(TEMP_FILE, INPUT_FILE)

print("Đã thêm index và ghi đè thành công:", INPUT_FILE)
```

# 2. Tổng hợp title vào cafef_news.json --> Tạo thành raw dataset hoàn chỉnh 

```python
import json

NEWS_FILE = "/Users/kittnguyen/Documents/DS201_Finance/data/raw/cafeF_news.json"
TITLES_FILE = "/Users/kittnguyen/Documents/DS201_Finance/data/raw/news_titles.jsonl"
OUTPUT_FILE = "/Users/kittnguyen/Documents/DS201_Finance/data/raw/cafef_news_raw.json"


# ---- 1. Load titles từ file .jsonl ----
titles = {}
with open(TITLES_FILE, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        obj = json.loads(line)
        titles[obj["_id"]] = obj["title"]


# ---- 2. Load dữ liệu cafef_news.json ----
with open(NEWS_FILE, "r", encoding="utf-8") as f:
    news_data = json.load(f)


# ---- 3. Thêm field title vào giữa post date và summary ----
new_data = []

for item in news_data:
    _id = item.get("_id")
    title = titles.get(_id, "")

    # tạo thứ tự key mới đúng như yêu cầu
    new_item = {}

    for key, value in item.items():
        if key == "post date":
            new_item[key] = value
            new_item["title"] = title   # chèn vào giữa post_date và summary
        else:
            new_item[key] = value

    new_data.append(new_item)


# ---- 4. Lưu file mới ----
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    json.dump(new_data, f, ensure_ascii=False, indent=4)

print(" Đã thêm title xong! File lưu tại:", OUTPUT_FILE)
```

Đây là file raw đầy đủ được sử dụng
