# Lọc ra các EVENT duy nhất

```python
import json

INPUT_FILE = "/Users/kittnguyen/Documents/DS201_Finance/data/labeling_data/output_tags/3_EVENT.json"            # file bạn muốn lọc
OUTPUT_FILE = "/Users/kittnguyen/Documents/DS201_Finance/data/labeling_data/output_tags/3_EVENT_unique.json"    # file output mới

# ---- 1. Load file ----
with open(INPUT_FILE, "r", encoding="utf-8") as f:
    data = json.load(f)

# ---- 2. Lấy evidence duy nhất ----
unique_evidences = []
seen = set()

for item in data:
    ev = item.get("evidence", "").strip()
    if ev and ev not in seen:
        seen.add(ev)
        unique_evidences.append({"sentence": ev})

# ---- 3. Lưu file mới ----
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    json.dump(unique_evidences, f, ensure_ascii=False, indent=4)

print("✔ Đã tạo file:", OUTPUT_FILE)
print("Số evidence duy nhất:", len(unique_evidences))
```

```python
import json
import re

def fix_sentence(s):
    # Tách chữ và số
    s = re.sub(r"(?<=[A-Za-zÀ-ỹ])(?=\d)", " ", s)   # chữ → số
    s = re.sub(r"(?<=\d)(?=[A-Za-zÀ-ỹ])", " ", s)   # số → chữ

    # Tách chữ thường → chữ hoa (bắt đầu tên/cụm mới)
    s = re.sub(r"(?<=[a-zà-ỹ])(?=[A-ZÀ-Ỹ])", " ", s)

    # Chuẩn hóa khoảng trắng
    s = re.sub(r"\s+", " ", s)
    s = re.sub(r"([A-Za-zÀ-ỹ])\s+([A-Za-zÀ-ỹ])", r"\1\2", s)
    return s.strip()
INPUT = "/Users/kittnguyen/Documents/DS201_Finance/data/labeling_data/output_tags/3_EVENT_unique.json"
OUTPUT = "/Users/kittnguyen/Documents/DS201_Finance/data/labeling_data/output_tags/3_EVENT_clean.json"

with open(INPUT, "r", encoding="utf-8") as f:
    data = json.load(f)

clean = [{"sentence": fix_sentence(item["sentence"])} for item in data]

with open(OUTPUT, "w", encoding="utf-8") as f:
    json.dump(clean, f, ensure_ascii=False, indent=4)

print("✔ Đã làm sạch:", OUTPUT)
```

```python
import json

INPUT_FILE = "/Users/kittnguyen/Documents/DS201_Finance/data/labeling_data/output_tags/3_EVENT_unique.json"
OUTPUT_FILE = "/Users/kittnguyen/Documents/DS201_Finance/data/labeling_data/output_tags/3_EVENT_clean.json"

def clean_sentence(s):
    if not isinstance(s, str):
        return s

    # 1. Remove all backslashes
    s = s.replace("\\", "")

    # 2. Strip leading spaces
    s = s.lstrip()

    # 3. Remove "-" if it is at the beginning
    if s.startswith("-"):
        s = s[1:].lstrip()

    return s

def clean_json_file():
    with open(INPUT_FILE, "r", encoding="utf-8") as f:
        data = json.load(f)

    for item in data:
        if "sentence" in item:
            item["sentence"] = clean_sentence(item["sentence"])

    with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=4)

    print("Done! File cleaned and saved to", OUTPUT_FILE)

clean_json_file()
```
