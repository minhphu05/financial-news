```python
import json
import os
from collections import defaultdict

INPUT_DIR = "/Users/kittnguyen/Documents/DS201_Finance/data/labeling_data/output_span_labels" # thư mục chứa 0.json → 11000.json
OUTPUT_DIR = "/Users/kittnguyen/Documents/DS201_Finance/data/labeling_data/output_tags"   # thư mục lưu 10 file theo tag
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ---- 1. Nhóm tất cả tag vào dict lớn ----
grouped = defaultdict(list)

# ---- 2. Lặp toàn bộ file từ 0.json → 11000.json ----
for i in range(0, 11021):
    path = os.path.join(INPUT_DIR, f"{i}.json")
    if not os.path.exists(path):
        continue

    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)  # list các item trong file

    for item in data:
        if not isinstance(item, dict):
            print("Lỗi ở file:", path)
            print("Item lỗi:", item)
            raise ValueError("Item không phải dict")
        label = item["label"]
        grouped[label].append(item)

print("Số lượng tag thu được:", {k: len(v) for k, v in grouped.items()})

# ---- 3. Lưu từng tag thành file JSON ----
for idx, (label, items) in enumerate(grouped.items()):
    filename = f"{idx}_{label}.json"
    filepath = os.path.join(OUTPUT_DIR, filename)

    with open(filepath, "w", encoding="utf-8") as f:
        json.dump(items, f, ensure_ascii=False, indent=4)

print("✔ Đã tạo file theo tag trong thư mục:", OUTPUT_DIR)
```
